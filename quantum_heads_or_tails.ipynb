```python
# Instalação e importação de bibliotecas

try:
    import cirq
    import matplotlib.pyplot as plt
except ImportError:
    print("Instalando cirq...")
    !pip install --quiet cirq
    import cirq

    print("Cirq Instalado!")
```

```python
# Criando um qubit para representar a moeda
qubit = cirq.GridQubit(0, 0)

# Criando o circuito quântico
circuit = cirq.Circuit(
    cirq.H(qubit),  # Aplicando a porta Hadamard para criar superposição (50% Cara, 50% Coroa)
    cirq.measure(qubit, key='coin')  # Medindo o qubit (0 = Cara, 1 = Coroa)
)

# Exibindo o circuito
print("Circuito Quântico para Cara ou Coroa:")
print(circuit)
```

```python
# Criando um simulador quântico
simulator = cirq.Simulator()

# Executando o circuito 1000 vezes para simular várias jogadas
repetitions = 1000
result = simulator.run(circuit, repetitions=repetitions)

# Obtendo a contagem dos resultados
counts = result.histogram(key='coin')

# Calculando porcentagens
cara_percent = (counts.get(0, 0) / repetitions) * 100
coroa_percent = (counts.get(1, 0) / repetitions) * 100

# Exibindo os resultados
print("\nResultados da Simulação de Cara ou Coroa:")
print(f"Cara (0): {counts.get(0, 0)} vezes ({cara_percent:.2f}%)")
print(f"Coroa (1): {counts.get(1, 0)} vezes ({coroa_percent:.2f}%)")
```

```python
# Criando um gráfico para visualizar os resultados
labels = ['Cara (0)', 'Coroa (1)']
values = [counts.get(0, 0), counts.get(1, 0)]

plt.figure(figsize=(6,4))
plt.bar(labels, values, color=['blue', 'red'])
plt.xlabel("Resultado")
plt.ylabel("Frequência")
plt.title("Distribuição dos Resultados - Cara ou Coroa Quântico")
plt.show()
```
