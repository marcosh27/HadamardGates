```python
import cirq
import string

simulator = cirq.Simulator()
qubits = [cirq.GridQubit(0, i) for i in range(5)]
circuit = cirq.Circuit()
circuit.append(cirq.H.on_each(*qubits))
circuit.append(cirq.measure(*qubits, key='result'))

def quantum_random_letter():
    while True:
        result = simulator.run(circuit, repetitions=1)
        bits = result.measurements['result'][0]
        value = int(''.join(str(bit) for bit in bits), 2)
        if value < 26:
            return string.ascii_lowercase[value]

def quantum_random_letters_batch(n):
    results = simulator.run(circuit, repetitions=n)
    letters = []
    for outcome in results.measurements['result']:
        value = int(''.join(str(bit) for bit in outcome), 2)
        if value < 26:
            letters.append(string.ascii_lowercase[value])
    return letters

def main():
    target = input("Digite a palavra que deseja encontrar: ").strip().lower()
    probability = (1/26) ** len(target)
    print(f"\nA probabilidade de gerar a palavra '{target}' é de 1 em {26**len(target)} ou aproximadamente {probability:.10f}\n")

    generated_text = ""
    total_letters = 0
    batch_size = 1000

    print(f"Iniciando a busca pela palavra '{target}'...")

    while target not in generated_text:
        letters = quantum_random_letters_batch(batch_size)
        generated_text += "".join(letters)
        total_letters += len(letters)

        if total_letters % 10000 < batch_size:
            print(f"{total_letters} letras geradas...")

    index = generated_text.find(target)
    margin = 20
    start = max(0, index - margin)
    end = index + len(target) + margin
    before = generated_text[start:index]
    word_found = generated_text[index:index+len(target)]
    after = generated_text[index+len(target):end]

    highlighted_word = f"\033[1;31m{word_found}\033[0m"

    print(f"\nPalavra '{target}' encontrada após gerar {total_letters} letras!")
    print("Trecho do texto gerado onde a palavra foi encontrada!")
    print(before + highlighted_word + after)

if __name__ == '__main__':
    main()
```
